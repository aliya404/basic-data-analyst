```python
# import file cvs pada folder di drive
import pandas as pd
import numpy as np

df = pd.read_excel('/content/drive/MyDrive/Aliya_DA_023/data_nilai_utbk_siswa_sma_pelita_harapan.xlsx')
df
```

```python
df_filled=df.fillna(0)
df_filled
```

```python
df=df.drop_duplicates(subset=['Nama', 'Jurusan'], keep='first')
df
```

```python
df['Matematika'] = df['Matematika'].fillna(df['Matematika'].mean())
df['Bahasa Indonesia'] = df['Bahasa Indonesia'].fillna(df['Bahasa Indonesia'].mean())
df['Bahasa Inggris'] = df['Bahasa Inggris'].fillna(df['Bahasa Inggris'].mean())
df['Fisika']= df['Fisika'].fillna(df['Fisika'].mean())
df['Kimia']= df['Kimia'].fillna(df['Kimia'].mean())
df['Ekonomi']= df['Ekonomi'].fillna(df['Ekonomi'].mean())
df['Sosiologi']= df['Sosiologi'].fillna(df['Sosiologi'].mean())
df
```

```python
df[['Matematika', 'Bahasa Indonesia', 'Bahasa Inggris', 'Fisika', 'Kimia', 'Ekonomi', 'Sosiologi']] = df[['Matematika', 'Bahasa Indonesia', 'Bahasa Inggris', 'Fisika', 'Kimia', 'Ekonomi', 'Sosiologi']].astype(float)
df
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[['Matematika', 'Bahasa Indonesia', 'Bahasa Inggris', 'Fisika', 'Kimia', 'Ekonomi', 'Sosiologi']] = scaler.fit_transform(df[['Matematika', 'Bahasa Indonesia', 'Bahasa Inggris', 'Fisika', 'Kimia', 'Ekonomi', 'Sosiologi']])
df
```

```python
Q1=df[['Matematika', 'Bahasa Indonesia', 'Bahasa Inggris', 'Fisika', 'Kimia', 'Ekonomi', 'Sosiologi']].quantile(0.25)
Q3=df[['Matematika', 'Bahasa Indonesia', 'Bahasa Inggris', 'Fisika', 'Kimia', 'Ekonomi', 'Sosiologi']].quantile(0.75)
IQR=Q3-Q1
print('IQR= ',IQR)
```
